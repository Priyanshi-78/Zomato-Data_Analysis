```python
cuisine_count = df.groupby("Cuisine ")["Item Name"].count().sort_values(ascending=False)

print(cuisine_count)
```

```python
cuisine_count.sort_values(ascending=False).head(1)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
x=cuisine_count.index,
y=cuisine_count.values,
color="blue"
)
plt.title("Cuisine with Highest Number of Menu Items")
plt.xlabel("Cuisine")
plt.ylabel("Number of Menu Items")
plt.xticks(rotation=90)
plt.savefig("Cuisine_with_Highest_Number_of_Menu_Items.png")
plt.show()
plt.close()
```

```python
top_restaurant = df.groupby("Restaurant Name")["Item Name"].count().sort_values(ascending=False).head(10)

print(top_restaurant)
```

```python
plt.figure(figsize=(12,6))
sns.barplot(
x=top_restaurant.index,
y=top_restaurant.values,
color="lightblue"
)
plt.title("Top 10 Restaurants with Most Menu Items")
plt.xlabel("Restaurant Name")
plt.ylabel("Number of Menu Items")
plt.xticks(rotation=90)
plt.savefig("Top_10_Restaurants.png")
plt.show()
plt.close()
```

```python
print(df["Prices"].describe())
```

```python
plt.figure(figsize=(12,6))
sns.histplot(
data=df,
x="Prices",
bins=30,
color="blue"
)
plt.title("Distribution of Menu Item Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.savefig("Menu_Item_Prices.png")
plt.show()
plt.close()
```

```python
top_items = df.sort_values(by="Votes", ascending=False).head(10)
print(top_items)
```

```python
df[["Item Name", "Votes"]]
```

```python
top_menu=df.groupby("Item Name")["Votes"].count().sort_values(ascending=False).head(10)
print(top_menu)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(
data=top_items,
x="Votes",
y="Item Name",
color="red"
)
plt.title("Top 10 Menu Items with Highest Votes")
plt.xlabel("Votes")
plt.ylabel("Menu Item")
plt.savefig("Top_10_Menu_Items.png")
plt.show()
plt.close()
```

```python
best_seller = df["Best Seller"].value_counts()
print(best_seller)
```

```python
plt.figure(figsize=(12,6))
sns.barplot(
x=best_seller.index,
y=best_seller.values,
color="skyblue"
)
plt.title("BESTSELLER, MUST TRY & CHEF'S SPECIAL")
plt.xlabel("Category")
plt.ylabel("Count")
plt.savefig("Best_Seller_Barplot.png")
plt.xticks(rotation=90)
plt.show()
plt.close()
```

```python
best_seller = df["Best Seller"].value_counts().head(5)
print(best_seller)
```

```python
best_seller = df["Best Seller"].value_counts().head(5)
plt.figure(figsize=(6,6))
plt.pie(
best_seller.values,
labels=best_seller.index,
autopct="%1.1f%%",
startangle=90
)
plt.title("Top 10 Best Seller Categories")
plt.savefig("Best_Seller_PieChart.png")
plt.show()
plt.close()
```

```python
city_restaurant = df.groupby("City")["Restaurant Name"].nunique().sort_values(ascending=False)
print(city_restaurant)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
x=city_restaurant.index,
y=city_restaurant.values,
color="blue"
)
plt.title("Restaurants Present in Each City")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=90)
plt.savefig("Restaurants_in_Each_City.png")
plt.show()
plt.close()
```

```python
avg_rating = df.groupby("City")["Dining Rating"].mean().sort_values(ascending=False)
print(avg_rating)
```

```python
plt.figure(figsize=(12,6))
sns.barplot(
x=avg_rating.index,
y=avg_rating.values,
color="blue"
)
plt.title("Average Dining Rating by City")
plt.xlabel("City")
plt.ylabel("Average Dining Rating")
plt.xticks(rotation=90)
plt.savefig("Average_Dining_Rating.png")
plt.show()
plt.close()
```

```python
menu_items = df.groupby("Cuisine ")["Item Name"].count().sort_values(ascending=False)
print(menu_items)
```

```python
plt.figure(figsize=(12,6))
sns.barplot(
x=menu_items.index,
y=menu_items.values,
color="blue"
)
plt.title("Menu Items Available in Each Cuisine")
plt.xlabel("Cuisine")
plt.ylabel("Number of Menu Items")
plt.xticks(rotation=90)
plt.savefig("Menu_Items_by_Cuisine.png")
plt.show()
plt.close()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
data=df,
x="Prices",
y="Votes",
color="blue"
)
plt.title("Relationship Between Price and Votes")
plt.xlabel("Price")
plt.ylabel("Votes")
plt.savefig("Price_vs_Votes.png")
plt.show()
plt.close()
```
