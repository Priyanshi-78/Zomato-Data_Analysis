```python
cuisine_count = df.groupby("Cuisine ")["Item Name"].count().sort_values(ascending=False)

print(cuisine_count)
```

```python
cuisine_count.sort_values(ascending=False).head(1)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
x=cuisine_count.index,
y=cuisine_count.values,
color="blue"
)
plt.title("Cuisine with Highest Number of Menu Items")
plt.xlabel("Cuisine")
plt.ylabel("Number of Menu Items")
plt.xticks(rotation=90)
plt.savefig("Cuisine_with_Highest_Number_of_Menu_Items.png")
plt.show()
plt.close()
```

```python
top_restaurant = df.groupby("Restaurant Name")["Item Name"].count().sort_values(ascending=False).head(10)

print(top_restaurant)
```

```python
plt.figure(figsize=(12,6))
sns.barplot(
x=top_restaurant.index,
y=top_restaurant.values,
color="lightblue"
)
plt.title("Top 10 Restaurants with Most Menu Items")
plt.xlabel("Restaurant Name")
plt.ylabel("Number of Menu Items")
plt.xticks(rotation=90)
plt.savefig("Top_10_Restaurants.png")
plt.show()
plt.close()
```

```python
print(df["Prices"].describe())
```

```python
plt.figure(figsize=(12,6))
sns.histplot(
data=df,
x="Prices",
bins=30,
color="blue"
)
plt.title("Distribution of Menu Item Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.savefig("Menu_Item_Prices.png")
plt.show()
plt.close()
```

```python
top_items = df.sort_values(by="Votes", ascending=False).head(10)
print(top_items)
```

```python
df[["Item Name", "Votes"]]
```

```python
top_menu=df.groupby("Item Name")["Votes"].count().sort_values(ascending=False).head(10)
print(top_menu)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(
data=top_items,
x="Votes",
y="Item Name",
color="red"
)
plt.title("Top 10 Menu Items with Highest Votes")
plt.xlabel("Votes")
plt.ylabel("Menu Item")
plt.savefig("Top_10_Menu_Items.png")
plt.show()
plt.close()
```

```python
best_seller = df["Best Seller"].value_counts()
print(best_seller)
```

```python
plt.figure(figsize=(12,6))
sns.barplot(
x=best_seller.index,
y=best_seller.values,
color="skyblue"
)
plt.title("BESTSELLER, MUST TRY & CHEF'S SPECIAL")
plt.xlabel("Category")
plt.ylabel("Count")
plt.savefig("Best_Seller_Barplot.png")
plt.xticks(rotation=90)
plt.show()
plt.close()
```
