```python
continent_count = df.groupby("continent")["team"].count()
print(continent_count)
```

```python
continent_count = continent_count.sort_values(ascending=False)
print(continent_count)
```

```python
plt.figure(figsize=(12,6))
sns.barplot(
   x=continent_count.index,
   y=continent_count.values,
   palette="Blues"
)
plt.title("Number of Teams by Continent")
plt.xlabel("Continent")
plt.ylabel("Number of Teams")
plt.show()
plt.close()
```

```python
host_team = df.groupby("is_host")["team"].count()
print(host_team)
```

```python
plt.figure(figsize=(8,8))
plt.pie(
   host_team.values,
   labels=["Non Host","Host"],
   autopct="%1.1f%%",
   startangle=90
)
plt.title("Host vs Non Host Teams")
plt.show()
plt.close()
```

```python
avg_rank = df.groupby("continent")["fifa_rank_pre_tournament"].mean()
print(avg_rank)
```

```python
avg_rank = avg_rank.sort_values()
print(avg_rank)
```

```python
plt.figure(figsize=(12,6))
sns.barplot(
   x=avg_rank.index,
   y=avg_rank.values,
   palette="deep"
)
plt.title("Average FIFA Ranking by Continent")
plt.xlabel("Continent")
plt.ylabel("Average FIFA Ranking")
plt.show()
plt.close()
```

```python
top_team = df.sort_values(by="squad_total_market_value_eur",ascending=False).head(10)
print(top_team[["team","squad_total_market_value_eur"]])
```

```python
plt.figure(figsize=(10,6))
sns.barplot(
   data=top_team,
   x="squad_total_market_value_eur",
   y="team",
   palette="viridis"
)
plt.title("Top 10 Teams by Squad Market Value")
plt.xlabel("Squad Market Value (EUR)")
plt.ylabel("Team")
plt.show()
plt.close()
```

```python
correlation = df[["fifa_rank_pre_tournament","fifa_points_pre_tournament"]].corr()
print(correlation)
```

```python
plt.figure(figsize=(12,6))
sns.scatterplot(
   data=df,
   x="fifa_rank_pre_tournament",
   y="fifa_points_pre_tournament",
   color="blue",
)
plt.title("FIFA Ranking vs FIFA Points")
plt.xlabel("FIFA Ranking")
plt.ylabel("FIFA Points")
plt.show()
plt.close()
```

```python
goals = df.sort_values(by="goals_scored_last_4y",ascending=False).head(10)
print(goals[["team","goals_scored_last_4y"]])
```

```python
plt.figure(figsize=(10,6))
sns.barplot(
    x=goals["team"],
    y=goals["goals_scored_last_4y"],
    palette="magma"
)
plt.xticks(rotation=90)
plt.title("Top teams by Goals Scored")
plt.xlabel("Team")
plt.ylabel("Goals Scored")
plt.show()
plt.close()
```

```python
continent_goals = df.groupby("continent")[["goals_scored_last_4y","goals_received_last_4y"]].sum()
print(continent_goals)
```

```python
continent_goals.plot(
    kind="bar",
    figsize=(10,6),
    color=["blue","skyblue"]
)
plt.title("Goals Scored vs Goals Received")
plt.xlabel("Continent")
plt.ylabel("Goals")
plt.show()
plt.close()
```

```python
titles = df.sort_values(by="world_cup_titles_before",ascending=False)
print(titles[["team","world_cup_titles_before"]])
```

```python
plt.figure(figsize=(10,6))
sns.barplot(
    data=titles,
    x="team",
    y="world_cup_titles_before",
    palette="coolwarm"
)
plt.xticks(rotation=90)
plt.title("World Cup Titles")
plt.xlabel("Team")
plt.ylabel("Number of Titles")
plt.show()
plt.close()
```
